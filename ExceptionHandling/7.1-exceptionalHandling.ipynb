## Understanding Exceptions

Exception handling in Python allows you to handle errors gracefully, without stopping the execution of the program. This lesson will cover the try, except, else, and finally blocks.

## What Are Exceptions?

Exceptions are events that disrupt the normal flow of a program's instructions during program execution. Common exceptions include:

- ZeroDivisionError: Dividing by zero.
- FileNotFoundError: File not found.
- ValueError: Invalid value.
- TypeError: Invalid type.

```python
a=10
# a=b #NameError

## Exception Handling
try:    
    a=b
except NameError as e:
    print("Variable b is not defined")
    print(e)
except Exception as e:
    print("An error occurred")
    print(e)
else:    print("No error occurred")
finally: print("This will always execute")
```

```python
try:
    result = 1/0
except ZeroDivisionError as e:
    print("Division by zero error occurred")
    print(e)
```

```python
try:
    num=int(input("Enter a number: "))
    result = 10/num
except ValueError as e:
    print("Invalid input. Please enter a valid number.")
    print(e)
except ZeroDivisionError as e:
    print("Division by zero error occurred")
    print(e)
except Exception as e:
    print("An error occurred")
    print(e)
else:    print("Result is:", result)
finally: print("This will always execute")
```

```python
## File Handling
try:
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError as e:
    print("File not found error occurred")
    print(e)
except Exception as e:
    print("An error occurred")
    print(e)
else:    print("File read successfully")
finally: 
    print("This will always execute")
    if 'file' in locals() or not file.closed:
        file.close()
        print("File closed")
```
