## The map() Function in Python

The map() function applies a given function to all the items in an input list (or any other iterable) and returns a map object (an iterator). This is particularly useful for transforming data in a list comprehensively

```python
def square(x):
    return x ** 2
numbers = [1, 2, 3, 4, 5]

squared_numbers = list(map(square, numbers))
print(squared_numbers)  # Output: [1, 4, 9, 16, 25]

# Using a lambda function with map()
numbers = [1, 2, 3, 4, 5]
squared = list(map(lambda x: x ** 2, numbers))
print(squared)  # Output: [1, 4, 9, 16, 25]
```

```python
# Mapping multiple iterables
numbers1 = [1, 2, 3]
numbers2 = [4, 5, 6]
summed = list(map(lambda x, y: x + y, numbers1, numbers2))
print(summed)  # Output: [5, 7, 9]
```

```python
# Map() to convert strings to uppercase
fruits = ['apple', 'banana', 'cherry']
uppercase_fruits = list(map(str.upper, fruits))
print(uppercase_fruits)  # Output: ['APPLE', 'BANANA', 'CH
```

```python
# Iterating over a dictionary with map()
person = {'name': 'Alice', 'age': 30, 'city': 'New York'}
keys = list(map(lambda x: x, person.keys()))
values = list(map(lambda x: x, person.values()))
print(keys)    # Output: ['name', 'age', 'city']
print(values)  # Output: ['Alice', 30, 'New York']
```

```python
def get_name(person):
    return person['name']
people = [
    {'name': 'Alice', 'age': 30},
    {'name': 'Bob', 'age': 25},
    {'name': 'Charlie', 'age': 35}
]
names = list(map(get_name, people))
print(names)  # Output: ['Alice', 'Bob', 'Charlie']
```
