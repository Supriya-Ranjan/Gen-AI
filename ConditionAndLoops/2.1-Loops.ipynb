```python
range(5) # - give range

## for loop

for i in range(5):
    print(i)

for i in range(1, 6): # last digit is excluded
    print(i)

for i in range(1, 10, 2):  # increment by 2
    print(i)

for i in range(10, 1, -1):  # inverse by 1, between 10 to 1, if we do (1, 10, -1) nothing will display, because 1-1 = 0 not coming between 1 to 10
    print(i)

for i in range(10, 1, -2):  # inverse by 2
    print(i)
```

```python
## strings

str="Krish Nair"

for i in str:
    print(i)
```

```python
## while loop

## the while loop continue to execute as long as the condition is True

count=0

while count<5:
    print(count)
    count=count+1
```

```python
## Loop control statements

#break
## The break statement is used to exit the loop when a certain condition is met. It can be used in both for and while loops.

for i in range(10):
    if i == 5:
        break  # Exit the loop when i is 5
    print(i)

## continue
## The continue statement is used to skip the current iteration of the loop and move to the next

for i in range(10):
    if i % 2 == 0:
        continue  # Skip the rest of the loop for even numbers
    print(i)  # This will only print odd numbers

# pass
## The pass statement is a placeholder that does nothing. It is used when a statement is syntactically required but you don't want to execute any code.

for i in range(10):
    if i == 5:
        pass  # Do nothing when i is 5
    print(i)
```

```python
## Nested loops
## A nested loop is a loop inside another loop. The inner loop will be executed for each
## iteration of the outer loop.

for i in range(5):
    for j in range(3):
        print(f"i: {i}, j: {j}")
```

```python
## Example -  Calculate the sum of the first n natural numbers using a for loop
n = 10
sum = 0
sum2 = 0
count = 1

while count <= n:
    sum += count
    count += 1

print(f"The sum of the first {n} natural numbers is: {sum}")

for i in range(1, n + 1):
    sum2 += i
print(f"The sum of the first {n} natural numbers is: {sum2}")
```

```python
## Example- Prime numbers between 1 to n
n = 20
print(f"Prime numbers between 1 and {n} are:")
for num in range(2, n + 1):
    is_prime = True
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            is_prime = False
            break
    if is_prime:
        print(num)
        
```
