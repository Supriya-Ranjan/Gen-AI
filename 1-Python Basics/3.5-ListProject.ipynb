Example 1. Manage a To-Do List
- Create a To Do List To keep Track Of Tasks

```python
to_do_list=["Buy groceries", "Clean the house", "Finish homework"]

print("To-Do List:")
for task in to_do_list:
    print("- " + task)

#Add a new task to the list
to_do_list.append("Call mom")
to_do_list.append("Pay bills")

print("\nUpdated To-Do List:")
for task in to_do_list:
    print("- " + task)

# Remove a task from the list
to_do_list.remove("Clean the house")

print("\nFinal To-Do List:")
for task in to_do_list:
    print("- " + task)

#Check if a task is in the list
task_to_check = "Finish homework"
if task_to_check in to_do_list:
    print(f"\n'{task_to_check}' is in the To-Do List.")
else:
    print(f"\n'{task_to_check}' is not in the To-Do List.")

# Task remaining in the list
print("\nRemaining Tasks:")
for task in to_do_list:
    print("- " + task)
```

Example 2. Organizing Student Grades
- Create a list To store and calculate average grades for students

```python
# Example 2: Organizing Student Grades
student_grades = [85, 92, 78, 90, 88]

#Add a new grade to the list
student_grades.append(95)

# Calculate the average grade
average_grade = sum(student_grades) / len(student_grades)
print(f"\nAverage Grade: {average_grade:.2f}")

# Find the highest and lowest grades
highest_grade = max(student_grades)
lowest_grade = min(student_grades)
print(f"Highest Grade: {highest_grade}")
print(f"Lowest Grade: {lowest_grade}")
```
