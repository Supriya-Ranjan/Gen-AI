```python
## if statements
age=20

if age>=18:
    print("you are allowed")

#else
if age>=18:
    print("you are allowed")
else:
    print("You are not allowed")

#elif - this allows you to check multiple conditions. It stands for "else if"

age1=20

if age1<13:
    print("You are less then 13")
elif age<18:
    print("You are between 13 and 18")
else:
    print("You are more then 18")

## Nested Conditional Statements
# You can  place one or more if, else, elif statements inside another

num=int(input("Enter a number: "))
if num>0:
    print("Positive num")
    if num%2==0:
        print("The number is even")
    else:
        print("The number is odd")
else:
    print("The number is 0 or negative")
```

```python
## Practical Example

## Determine if a year is a leap year using nested condition statement

year=int(input("Enter a year: "))

if year%4==0:
    if year%100==0:
        if year%400==0:
            print(year, "is a leap year")
        else:
            print(year, "is not a leap year")
    else:
            print(year, "is a leap year")
else:
            print(year, "is not a leap year")
```
