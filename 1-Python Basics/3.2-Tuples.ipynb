# Explanation
Tuples are ordered collections of items that are immutable. They are similar to lists, but their immutability makes them different.

```python
## creating a tuple
empty_tuple=()
print(empty_tuple)
print(type(empty_tuple))

tpl=tuple()
lst=list()

number=tuple([1,2,3,4,5,6])
list((1,2,3,4,5,6))

mixed_tuple=(1, "Hello", "233", 3.41)

#indexing and slicing similar to list
```

```python
## Concatenation
number=tuple([1,2,3,4,5,6])
mixed_tuple=(1, "Hello", "233", 3.41)
concatenation = number + mixed_tuple
print(concatenation)

mixed_tuple * 3 # all the element repeated by 3 times same in list
# (1, 'Hello', '233', 3.41, 1, 'Hello', '233', 3.41, 1, 'Hello', '233', 3.41)
```

```python
## Immutable nature of tuple - can not change the element once assign, can change after converting into list

# number[1] = "Krish" #error
```

```python
# Tuples Method

print(number.count(1))
print(number.index(3)) # return 1st index of value
```

```python
## packing and unpacking tuple
packed_tuple=1, "Hello", 3.14
print(packed_tuple)

a,b,c=packed_tuple
print(a)
print(b)
print(c)

# unpacking with *
numbers=[1,2,3,4,5,6]
first,*middle,last=numbers
print(first) # 1
print(middle) # [2,3,4,5]
print(last) # 6
```

```python
## Nested Tuple
tuple=((1,2,3,4), (5,6,7,8), (1, "hello", 20.45))
print(tuple[0][2]) # same in list

for sub_tuple in tuple:
    for item in sub_tuple:
        print(item, end=" ")
    print()
```

# Conclusion
Tuples are versatile and useful in many real-world scenarios where a fixed collection of items is required. They are commonly used for representing fixed data, returning multiple values from a function, and as dictionary keys. Understanding how to use tuples will help you write more robust and readable Python code.
