```python
## create a new directory
import os
new_directory = 'new_folder'
try:
    os.mkdir(new_directory)
    print(f"Directory '{new_directory}' created successfully.")
except FileExistsError:
    print(f"Directory '{new_directory}' already exists.")

```

```python
## List all files and directories in the current directory

items = os.listdir('.')
print("Files and Directories in the current directory:")
for item in items:
    print(item)

current_directory = os.getcwd()
print(f"Current Directory: {current_directory}")
items = os.listdir(current_directory)
print("Files and Directories in the current directory:")
for item in items:
    print(item)
```

```python
# joining paths
file_name = 'example.txt'
full_path = os.path.join(os.getcwd(),current_directory, file_name)
print(f"Full path to the file: {full_path}")
```

```python
path="example.txt"
if os.path.exists(path):
    print(f"The file '{path}' exists.")
else:
    print(f"The file '{path}' does not exist.")
```

```python
# checking if a path is a file or a directory
if os.path.isfile(path):
    print(f"'{path}' is a file.")
elif os.path.isdir(path):
    print(f"'{path}' is a directory.")
```

```python
# getting the size of a file
file_size = os.path.getsize(path)
print(f"The size of the file '{path}' is {file_size} bytes.")
```

```python
#getting the last modified time of a file
last_modified_time = os.path.getmtime(path)
print(f"The last modified time of the file '{path}' is: {last_modified_time}")
```

```python
#getting the absolute path of a file
absolute_path = os.path.abspath(path)
print(f"The absolute path of the file '{path}' is: {absolute_path}")
```
