# Inheritance In Python

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers different types of inheritance, demonstrating how to create and use them in Python.

```python
## Inheritance
## Parent class

class Car:
    def __init__(self, make, model, window, door_status):
        self.make = make
        self.model = model
        self.window = window
        self.door_status = door_status

    def display_info(self):
        print(f"Car Make: {self.make}, Model: {self.model}, Window: {self.window}, Door Status: {self.door_status}")

car1 = Car("Toyota", "Camry", "Open", "Closed")
car1.display_info()

## Child class inheriting from Car
class ElectricCar(Car):
    def __init__(self, make, model, window, door_status, battery_capacity):
        # Call the constructor of the parent class
        super().__init__(make, model, window, door_status)
        self.battery_capacity = battery_capacity

    def display_battery_info(self):
        print(f"Battery Capacity: {self.battery_capacity} kWh")

electric_car1 = ElectricCar("Tesla", "Model S", "Closed", "Closed", 100)
electric_car1.display_info()  # Inherited method from Car
electric_car1.display_battery_info()  # Method from ElectricCar
```

```python
# Multiple Inheritance
# where a class can inherit from more than one parent class

class Animal:
    def eat(self):
        print("Eating...")

class Mammal(Animal):
    def walk(self):
        print("Walking on all fours...")

class Bird(Animal):
    def fly(self):
        print("Flying in the sky...")
class Bat(Mammal, Bird):
    def display_info(self):
        print("I am a Bat, I can walk and fly.")
```

```python
class baseClass1:
    def __init__(self, name):
        self.name = name

    def display_name(self):
        print(f"Name from base class: {self.name}")

class baseClass2:
    def __init__(self, age):
        self.age = age

    def display_age(self):
        print(f"Age from base class: {self.age}")

class DerivedClass(baseClass1, baseClass2):
    def __init__(self, name, age):
        baseClass1.__init__(self, name)  # Initialize baseClass1
        baseClass2.__init__(self, age)   # Initialize baseClass2

    def display_info(self):
        self.display_name()  # Call method from baseClass1
        self.display_age()   # Call method from baseClass2
        print("This is the derived class.")

derived_object = DerivedClass("Alice", 30)
derived_object.display_info()
```
