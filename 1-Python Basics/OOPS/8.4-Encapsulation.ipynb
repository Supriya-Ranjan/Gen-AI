# Encapsulation

Encapsulation and abstraction are two fundamental principles of Object-Oriented Programming (OOP) that help in designing robust, maintainable, and reusable code. Encapsulation involves bundling data and methods that operate on the data within a single unit, while abstraction involves hiding the complex implementation details and exposing only the necessary features.

# Encapsulation

Encapsulation is the concept of wrapping data (variables) and methods (functions) together as a single unit. It restricts direct access to some of the object's components, which is a means of preventing accidental interference and misuse of the data.

```python
# Encapsulation with Getter and Setter Methods
# Encapsulation can be achieved in Python using getter and setter methods. These methods allow controlled access to the attributes of a class, ensuring that the internal state of an object is protected from unauthorized access and modification.
# Public, Protected, and Private Attributes
# In Python, there are three types of attributes based on their accessibility: public, protected, and private. Public attributes can be accessed from anywhere, protected attributes are intended to be accessed only within the class and its subclasses, and private attributes are intended to be accessed only within the class itself.

class Person:
    def __init__(self, name, age):
        self.name = name  # Public attribute
        self._age = age   # Protected attribute with a single underscore
        self.__ssn = "123-45-6789"  # Private attribute with double underscores

    def get_ssn(self):
        return self.__ssn

    def set_ssn(self, ssn):
        self.__ssn = ssn

class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id

    def get_employee_info(self, age):
        self._age = age  # Modifying protected attribute from subclass (not recommended)
        return f"Name: {self.name}, Age: {self._age}, Employee ID: {self.employee_id}"

person = Person("Alice", 30)
print(person.name)  # Accessing public attribute
print(person._age)  # Accessing protected attribute (not recommended) should be accessed within the class or subclass
print(person.get_ssn())  # Accessing private attribute through getter method
person.set_ssn("987-65-4321")  # Modifying private attribute through setter method
print(person.get_ssn())  # Accessing modified private attribute through getter method

employee = Employee("Bob", 25, "E123")
print(employee.name)  # Accessing public attribute from subclass
print(employee._age)  # Accessing protected attribute from subclass (not recommended)
print(employee.get_ssn())  # Accessing private attribute through getter method
employee.set_ssn("987-65-4321")  # Modifying private attribute through setter method
print(employee.get_ssn())  # Accessing modified private attribute through getter method
print(employee.get_employee_info(23)) # Modifying protected attribute from subclass (not recommended) and accessing employee info
```
