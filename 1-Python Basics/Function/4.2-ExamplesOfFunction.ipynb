```python
# Convert temperature from Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    """Convert Celsius to Fahrenheit."""
    return (celsius * 9/5) + 32

print(celsius_to_fahrenheit(0))   # 32.0
print(celsius_to_fahrenheit(100)) # 212.0
```

```python
# Password Strength Checker
def check_password_strength(password):
    """Check the strength of a password."""
    if len(password) < 6:
        return "Weak"
    elif len(password) < 12:
        return "Moderate"
    else:
        if any(char.isdigit() for char in password) and any(char.isupper() for char in password) and any(char.islower() for char in password) and any(char in "!@#$%^&*()-_=+[]{}|;:'\",.<>?/" for char in password):
            return "Not Strong"
        return "Strong"
print(check_password_strength("abc"))          # Weak
print(check_password_strength("password123")) # Moderate
print(check_password_strength("P@ssw0rd657tf"))   # Not Strong
print(check_password_strength("P@ssw0rd!2024")) # Strong
```

```python
# Calculate the total price of items in a shopping cart
def calculate_total_price(prices):
    """Calculate the total price of items in a shopping cart."""
    total = sum(prices)
    return total

print(calculate_total_price([19.99, 5.49, 3.50])) # 28.98
```

```python
# Is Palindrome Checker
def is_palindrome(s):
    """Check if a string is a palindrome."""
    cleaned = ''.join(s.lower().split())
    return cleaned == cleaned[::-1]

print(is_palindrome("A man a plan a canal Panama")) # True
print(is_palindrome("Not a palindrome"))             # False
```

```python
# A Function to Read a File and Count the Number of Words and characters in it
def count_words_in_file(file_path):
    """Count the number of words in a file."""
    with open(file_path, 'r') as file:
        text = file.read()
        words = text.split()
        return len(words), len(text)

# Example usage:
print(count_words_in_file('example.txt')) # Output will depend on the content of example.txt

# count word frequency in a file
def word_frequency(file_path):
    """Count the frequency of each word in a file."""
    with open(file_path, 'r') as file:
        text = file.read().lower()
        words = text.split()
        print(words)
        frequency = {}
        for word in words:
            frequency[word] = frequency.get(word, 0) + 1
        return frequency

# Example usage:
print(word_frequency('example.txt')) # Output will depend on the content of example.txt
```

```python
# validate email address
import re
def validate_email(email):
    """Validate an email address."""
    pattern = r'^[\w\.-]+@[\w\.-]+\.\w+$'
    if re.match(pattern, email):
        return "Valid"
    else:
        return "Invalid"
    
print(validate_email("test@example.com"))  # Valid
print(validate_email("invalid-email"))      # Invalid
```
