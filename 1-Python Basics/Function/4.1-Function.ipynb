# Function in Python

Definition:

A function is a block of code that performs a specific task, improving readability and reusability.

```python
## syntax

def function_name(parameters):
    """Docstring: Describe what the function does."""
    # code block
    return "value"

def greet(name):
    """Greet the user by name."""
    return f"Hello, {name}!"

print(greet("Alice"))
```

```python
## why use functions
# 1. Reusability: Write once, use multiple times.
# 2. Modularity: Break down complex problems into smaller, manageable pieces.
# 3. Readability: Clear structure and intent.
# 4. Maintainability: Easier to update and debug code.

# Checking if a number is prime
def is_prime(n):
    """Check if a number is prime."""
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

print(is_prime(7))  # True
print(is_prime(10)) # False
```

```python
# functions with multiple parameters
def calculate_area(length, width):
    """Calculate the area of a rectangle."""
    return length * width

print(calculate_area(5, 10))  # 50
```

```python
# Default parameters
def greet(name="Guest"):
    """Greet the user by name, or as Guest if no name is provided."""
    return f"Hello, {name}!"
print(greet())          # Hello, Guest!
print(greet("Alice"))   # Hello, Alice!
```

```python
# Variable Length Arguments
# Positional Variable Length Arguments
# *args allows you to pass a variable number of non-keyword arguments to a function.
def sum_all(*args):
    """Sum all provided numbers."""
    return sum(args)

print(sum_all(1, 2, 3))  # 6
print(sum_all(5, 10, 15)) # 30
```

```python
# Keyword Variable Length Arguments
# **kwargs allows you to pass a variable number of keyword arguments to a function.
def print_info(**kwargs):
    """Print key-value pairs provided as keyword arguments."""
    for key, value in kwargs.items():
        print(f"{key}: {value}")
        
print_info(name="Alice", age=30, city="New York")
```

```python
# combined use of *args and **kwargs
def combined_example(*args, **kwargs):
    """Demonstrate combined use of *args and **kwargs."""
    print("Positional arguments (args):", args)
    print("Keyword arguments (kwargs):", kwargs)
    print(kwargs.get("name", "No name provided"))

combined_example(1, 2, 3, name="Alice", age=30)
```

```python
# Returning Multiple Values
def calculate_stats(numbers):
    """Calculate and return the sum, average, and count of a list of numbers."""
    total = sum(numbers)
    count = len(numbers)
    average = total / count if count > 0 else 0
    return total, average, count
stats = calculate_stats([1, 2, 3, 4, 5])
print(f"Total: {stats[0]}, Average: {stats[1]}, Count: {stats[2]}")
```

```python
# Variable scope
def outer_function():
    outer_var = "I am outside!"
    
    def inner_function():
        inner_var = "I am inside!"
        print(outer_var)  # Accessing outer variable
        print(inner_var)  # Accessing inner variable
    
    inner_function()
    # print(inner_var)  # This would cause an error (inner_var is not accessible here)
outer_function()
```
