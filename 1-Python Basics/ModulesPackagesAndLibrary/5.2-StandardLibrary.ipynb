```python
# Data serialization using the json module
import json
data = {
    "name": "Alice",
    "age": 30,
    "city": "New York"
}
# Serialize to JSON string
json_string = json.dumps(data)
print(json_string)
print(type(json_string))

parsed_data = json.loads(json_string)
print(parsed_data)
print(type(parsed_data))
```

```python
# csv module for handling CSV files
import csv
# Writing to a CSV file
with open('data.csv', 'w', newline='') as csvFile:
    writer = csv.writer(csvFile)
    writer.writerow(['Name', 'Age', 'City'])
    writer.writerow(['Alice', 30, 'New York'])
    writer.writerow(['Bob', 25, 'Los Angeles'])

# Reading from a CSV file
with open('data.csv', 'r') as csvFile:
    reader = csv.reader(csvFile)
    for row in reader:
        print(row)
```

```python
## datetime module for handling dates and times
from datetime import datetime, timedelta

now = datetime.now()
print("Current date and time:", now)
print("Year:", now.year)
print("Month:", now.month)
print("Day:", now.day)
print("Hour:", now.hour)
print("Minute:", now.minute)
print("Second:", now.second)

print("Formatted date:", now.strftime("%Y-%m-%d %H:%M:%S"))
# timedelta for date arithmetic
one_week = timedelta(weeks=1)
next_week = now + one_week
print("Date one week from now:", next_week)
print("Date one week ago:", now - one_week)
```

```python
# time module for time-related functions
import time
print("Current time in seconds since the epoch:", time.time())
print("Current local time:", time.localtime())
print("Current time formatted:", time.strftime("%Y-%m-%d %H:%M:%S", time.localtime()))
print("Sleeping for 2 seconds...")
time.sleep(2)
print("Awake now!")
```

```python
# Regular expressions using the re module
import re
text = "The rain in Spain stays mainly in the plain."
# Find all words that start with 's'
matches = re.findall(r'\bs\w*', text)
print(matches)

pattern = r'\b\w+ain\b'
matches = re.findall(pattern, text)
print(matches)
```
