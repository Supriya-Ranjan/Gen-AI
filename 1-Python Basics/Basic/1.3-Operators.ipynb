```python
# Arithmetic Operator
a = 10
b = 5

add_result=a+b #addition
sub_result=a-b #subtraction
multi_result=a*b #multiplication
div_result=a/b #division
floor_div_result=a//b ## floor division
modulus_result=a%b #modulus operation
exponent_result=a**b ## Exponentiation

print(add_result)
print(sub_result)
print(multi_result)
print(div_result)
print(floor_div_result)
print(modulus_result)
```

```python
## Comparison Operators
## == Equal To Operator
a="Krish"
b="Krish"
a==b

## Not Equal to !=
a="Krish"
b="krish"

a==b

num1 = 45
num2 = 55

num1>num2 # greater than
num1<num2 # less than
num1>=num2 # greater than equal to
num1<=num2 # less than qual to
```

```python
## Logical Operators

X=True
Y=True

result = X and Y #both should true
result1= X or Y #one of them is true
result2 = not X #inverse

print(result)
print(result1)
print(result2)
```
