sets are tools used to handling unique items and doing fast operations like unions and intersections.

# A set is:

- Unordered
- No duplicate values
- Mutable (you can add/remove items)

```python
# Direct
s = {1, 2, 3}

# From list (removes duplicates)
s = set([1, 2, 2, 3])  # {1, 2, 3}

# Empty set (important!)
s = set()   # NOT {} → that's a dict
```

```python
s = {1, 2}

s.add(3)        # {1, 2, 3}
s.remove(2)     # {1, 3} (error if not found)
s.discard(10)   # no error
s.pop()         # removes random element
```

```python
if 1 in s:
    print("Yes")

a = {1, 2, 3}
b = {3, 4, 5}

# Union (combine)
a | b   # {1, 2, 3, 4, 5}

# Intersection (common)
a & b   # {3}

# Difference
a - b   # {1, 2}

# Symmetric difference
a ^ b   # {1, 2, 4, 5}
```

```python
# Remove duplicates from list
nums = [1, 2, 2, 3]
unique = list(set(nums))

#Check if subset
a = {1, 2}
b = {1, 2, 3}

a.issubset(b)   # True
```

When to use sets

Use sets when you need:

- Remove duplicates
- Fast lookup
- Compare collections
- Find common/unique items
