# Classes and Object 

Object-Oriented Programming (OOPS) is a programming paradigm that uses "objects" to design applications and computer programs. OOP allows for modelling real-world scenarios using classes and objects. This lesson covers the basic of creating classes and objects, including instance variables and methods.

```python
# A class is a blueprint for creating objects. It defines a set of attributes and methods that the created objects will have. An object is an instance of a class, which means it is created based on the structure defined by the class.
# In Python, you can define a class using the `class` keyword. Here's a simple example of a class definition and how to create an object from it:

class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print("Woof!")

# Creating an object of the Dog class
my_dog = Dog("Buddy", 3)

dir(my_dog)
```

```python
# Instance variables are variables that are specific to each instance of a class. They are defined within the `__init__` method and are accessed using the `self` keyword. In the example above, `name` and `age` are instance variables.
# Methods are functions that are defined within a class and can be called on objects of that class. In the example above, `bark` is a method that can be called on the `my_dog` object.

class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        print("Woof!") 

# Creating an object of the Dog class
my_dog = Dog("Buddy", 3)

my_dog.bark()  # Output: Woof!
print(my_dog.name)  # Output: Buddy
print(my_dog.age)   # Output: 3
my_dog.color = "Brown"  # Adding a new instance variable to the my_dog object
print(my_dog.color)  # Output: Brown

my_dog2 = Dog("Max", 5)
my_dog2.color = "Black"
print(my_dog2.color)  # Output: Black
```

```python
#  Instance variables and Methods

class Dog:
    ## constructor method to initialize instance variables
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Method to make the dog bark
    def bark(self):
        print(f"{self.name} says Woof!")

dog1 = Dog("Buddy", 3)
dog1.bark()  # Output: Buddy says Woof!
print(dog1.name)  # Output: Buddy
print(dog1.age)   # Output: 3
```

```python
## Modeling a Bank Account using OOP

## Defining a BankAccount class
class BankAccount:
    def __init__(self, account_number, balance=0):
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited ${amount}. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        elif amount <= 0:
            print("Withdrawal amount must be positive.")
        else:
            self.balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self.balance}")
    def check_balance(self):
        print(f"Account Number: {self.account_number}, Balance: ${self.balance}")
# Creating a bank account object
account1 = BankAccount("123456789", 1000)
account1.check_balance()  # Output: Account Number: 123456789, Balance: $1000
account1.deposit(500)    # Output: Deposited $500. New balance: $1500
account1.withdraw(200)   # Output: Withdrew $200. New balance: $1300
account1.withdraw(1500)  # Output: Insufficient funds.
account1.deposit(-100)  # Output: Deposit amount must be positive.
account1.withdraw(-50)  # Output: Withdrawal amount must be positive.
account1.check_balance()  # Output: Account Number: 123456789, Balance: $1300
```
